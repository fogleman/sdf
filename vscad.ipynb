```python
# imports
from sdf import *
import pyvista as pv
```

```python
# create shape

f = sphere(1) & box(1.5)

c = cylinder(0.5)
f -= c.orient(X) | c.orient(Y) | c.orient(Z)
```

```python
# display and output variables
window_size = [1600, 900]
stl_samples = 2**26
file_name = "object.stl"

# create plotter
pv.global_theme = pv.themes.DarkTheme()
plotter = pv.Plotter(window_size=window_size)
```

```python
# save to stl
f.save(file_name, samples=stl_samples)

# read and display stl
mesh = pv.read(file_name)
plotter.add_mesh(mesh, color="lightblue")
plotter.show()
```
